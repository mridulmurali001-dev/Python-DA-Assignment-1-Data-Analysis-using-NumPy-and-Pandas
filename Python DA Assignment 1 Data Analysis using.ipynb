```python
import numpy as np      # NumPy: fast numerical arrays and math
import pandas as pd     # Pandas: Series and DataFrame for tabular data

# PART 1: NUMPY ARRAY OPERATIONS

print("=" * 60)
print("PART 1: NUMPY ARRAY OPERATIONS")
print("=" * 60)

#  Step 1: Create a 1D NumPy array
# np.array() converts a Python list into a NumPy array.
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print("Week 1 temperatures (C):", temperatures_w1)

# Step 2: Inspection and properties
print("\nShape          :", temperatures_w1.shape)   # (7,) -> 1D array with 7 elements
print("Data type      :", temperatures_w1.dtype)    # float64 because values have decimals
print("No. of elements:", temperatures_w1.size)     # total count of elements = 7

#  Step 3: Array operations
# NumPy applies arithmetic to every element at once,
# so no loop is needed.
temperatures_f = (temperatures_w1 * 9 / 5) + 32
print("\nWeek 1 temperatures (F):", temperatures_f)

# Built-in aggregate functions
print("Maximum:", temperatures_w1.max())                 # highest temperature
print("Minimum:", temperatures_w1.min())                 # lowest temperature
print("Mean   :", round(temperatures_w1.mean(), 2))      # average, rounded to 2 decimals

# Step 4: Slicing and indexing
# Slice syntax: array[start:stop]
print("\nFirst three days :", temperatures_w1[:3])    # indices 0,1,2
print("Weekend (last 2) :", temperatures_w1[-2:])     # negative index counts from the end
print("Middle three days:", temperatures_w1[2:5])     # indices 2,3,4 (Day 3 to Day 5)

#  Step 5: Create a 2D array
# A list of lists becomes a 2D array: each inner list is one row
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],   # Week 1
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]    # Week 2
])
print("\n2D temperatures array:\n", temperatures)

# Step 6: Inspect and slice the 2D array
print("\nShape          :", temperatures.shape)   # (2, 7) -> 2 rows, 7 columns
print("Data type      :", temperatures.dtype)    # float64
print("Total elements :", temperatures.size)     # 2 x 7 = 14

# array[row, column] -> row index first, column index second
print("\nWeek 1 temperatures:", temperatures[0])      # row 0 (all columns)
print("Week 2 temperatures:", temperatures[1])        # row 1 (all columns)

# ':' means "all rows"; '-2:' means "last two columns" -> weekends of both weeks
print("Weekend temperatures (both weeks):\n", temperatures[:, -2:])
print("Week 1 weekend:", temperatures[0, -2:])
print("Week 2 weekend:", temperatures[1, -2:])


# PART 2: PANDAS SERIES

print("\n" + "=" * 60)
print("PART 2: PANDAS SERIES")
print("=" * 60)

#  Step 1: Create a Series with custom indices
# data = the values, index = custom labels instead of the default 0,1,2...
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)
print("Marks Series:\n", marks)

#  Step 2: Indexing and slicing
# Integer position access (position 0 = first element)
print("\nMark of 1st rank (marks[0]) :", marks.iloc[0])


# loc: access by LABEL. Label slices INCLUDE the end label.
print("Top 3 ranks using loc:\n", marks.loc['Rank1':'Rank3'])

# iloc: access by integer POSITION. 3rd student = position 2.
print("\n3rd rank mark using iloc:", marks.iloc[2])

# Boolean mask: the condition creates True/False for each element;
# passing it back into the Series keeps only the True ones.
print("\nRanks with marks > 90:\n", marks[marks > 90])

# Step 3: Manipulating the Series
# Modify a value by label
marks['Rank1'] = 100
print("\nAfter modifying Rank1 to 100:\n", marks)

# drop() returns a NEW Series without the given label, so we reassign it.
marks = marks.drop('Rank5')
print("\nAfter removing Rank5:\n", marks)

# Vectorized division: every mark is divided by 10 to get CGPA.
cgpa = marks / 10
print("\nCGPA:\n", cgpa)

# PART 3: PANDAS DATAFRAME

print("\n" + "=" * 60)
print("PART 3: PANDAS DATAFRAME")
print("=" * 60)

#  Step 1: Create the DataFrame
# A dictionary: keys become column names, lists become column values.
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture',
                        'Clothing', 'Electronics', 'Furniture', 'Clothing',
                        'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West',
               'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

# Step 2: Data exploration
print("Full DataFrame:\n", transactions)
print("\nHead (first 5 rows):\n", transactions.head())   # default = 5 rows
print("\nTail (last 5 rows):\n", transactions.tail())    # default = 5 rows
print("\nShape:", transactions.shape)                    # (rows, columns) = (10, 4)
print("Column names:", list(transactions.columns))
print("\nData types:\n", transactions.dtypes)
print("\nBasic info:")
transactions.info()                                      # prints non-null counts and dtypes

# Select multiple columns: pass a LIST of column names (double brackets)
print("\nProductCategory and Amount only:\n",
      transactions[['ProductCategory', 'Amount']])

# iloc[rows, columns]: ':' = all rows, '-3:' = last 3 columns
print("\nLast 3 columns (iloc):\n", transactions.iloc[:, -3:])
# Same result using loc with column labels:
print("\nLast 3 columns (loc):\n",
      transactions.loc[:, ['ProductCategory', 'Region', 'Amount']])

# Combine conditions with & (AND); each condition MUST be in parentheses.
north_high = transactions[(transactions['Region'] == 'North') &
                          (transactions['Amount'] > 200)]
print("\nNorth region with Amount > 200:\n", north_high)

# value_counts(): how many times each category appears
print("\nProductCategory value counts:\n",
      transactions['ProductCategory'].value_counts())

# unique(): distinct values in a column
print("\nUnique regions:", transactions['Region'].unique())

# groupby(): split by Region, then apply mean() to the Amount of each group
print("\nMean amount per region:\n",
      transactions.groupby('Region')['Amount'].mean())

#  Step 3: Manipulating the DataFrame
# loc[row_condition, column]: set Amount to 165 where TransactionID is 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print("\nAfter updating TransactionID 102 to 165:\n", transactions)

# New column: 10% of each Amount (vectorized multiplication)
transactions['Discount'] = transactions['Amount'] * 0.10
print("\nAfter adding Discount column:\n", transactions)

# Remove a row: find its index with a condition, then drop it.
# (Keep rows where TransactionID is NOT 109 -- simple and readable.)
transactions = transactions[transactions['TransactionID'] != 109]
print("\nAfter removing TransactionID 109:\n", transactions)

# Remove a column: axis=1 means "column" (axis=0 would mean "row")
transactions = transactions.drop('Discount', axis=1)
print("\nAfter deleting the Discount column:\n", transactions)
```
